```python
import numpy as np

n = len(ranks)

for config in ["A", "B"]:
    rank = ranks[f"rank_{config}"]

    recall_1 = (rank <= 1).mean()
    recall_3 = (rank <= 3).mean()
    recall_5 = (rank <= 5).mean()

    reciprocal_rank = np.where(rank <= 10, 1 / rank, 0)
    mrr_10 = reciprocal_rank.mean()

    print(f"Config {config}")
    print(f"Recall@1: {recall_1:.4f}")
    print(f"Recall@3: {recall_3:.4f}")
    print(f"Recall@5: {recall_5:.4f}")
    print(f"MRR@10:   {mrr_10:.4f}")
    print()
```

```python
from statsmodels.stats.proportion import proportion_confint

n = 60
candidates = [
    ("A, k=3", 41),
    ("A, k=5", 46),
    ("B, k=3", 36),
    ("B, k=5", 50)]
for name, x in candidates:
    recall = x / n
    lower, upper = proportion_confint(x, n, alpha=0.05, method="wilson")
    print(name)
    print("Recall:", round(recall, 4))
    print("Lower:", round(lower, 4))
    print("Upper:", round(upper, 4))
    print()
```

```python

# Part 1c: Annual model cost in USD

requests_per_day = 7500
days_per_year = 365
system_prompt_tokens = 245
input_price = 3
output_price = 18


data = {
    "EN": {
        "share": 0.35,
        "question": 40,
        "answer": 90,
        "chunk_A": 268,
        "chunk_B": 69
    },
    "RU": {
        "share": 0.27,
        "question": 57,
        "answer": 129,
        "chunk_A": 383,
        "chunk_B": 99
    },
    "KK": {
        "share": 0.38,
        "question": 141,
        "answer": 318,
        "chunk_A": 946,
        "chunk_B": 244
    }
}

def annual_cost(config, k):
    daily_input_tokens = 0
    daily_output_tokens = 0

    for lang, values in data.items():
        requests = requests_per_day * values["share"]
        chunk_tokens = values[f"chunk_{config}"]

        input_per_request = (
            system_prompt_tokens
            + values["question"]
            + k * chunk_tokens
        )

        output_per_request = values["answer"]

        daily_input_tokens += requests * input_per_request
        daily_output_tokens += requests * output_per_request


    annual_input_tokens = daily_input_tokens * days_per_year
    annual_output_tokens = daily_output_tokens * days_per_year

    cost = (
        annual_input_tokens * input_price
        + annual_output_tokens * output_price
    ) / 1_000_000

    return cost


results = {
    "A, k=3": annual_cost("A", 3),
    "A, k=5": annual_cost("A", 5),
    "B, k=3": annual_cost("B", 3),
    "B, k=5": annual_cost("B", 5)
}


for candidate, cost in results.items():
    print(f"{candidate}: ${cost:,.2f}")
```
