```python
from statsmodels.stats.proportion import proportion_confint

n = 60
candidates = [
    ("A, k=3", 41),
    ("A, k=5", 46),
    ("B, k=3", 36),
    ("B, k=5", 50)]
for name, x in candidates:
    recall = x / n
    lower, upper = proportion_confint(x, n, alpha=0.05, method="wilson")
    print(name)
    print("Recall:", round(recall, 4))
    print("Lower:", round(lower, 4))
    print("Upper:", round(upper, 4))
    print()
```

```python

# Part 1c: Annual model cost in USD

requests_per_day = 11000
days_per_year = 365
system_prompt_tokens = 195
input_price = 1
output_price = 6

data = {
    "EN": {
        "share": 0.44,
        "question": 28,
        "answer": 131,
        "chunk_A": 274,
        "chunk_B": 76
    },
    "RU": {
        "share": 0.41,
        "question": 51,
        "answer": 237,
        "chunk_A": 496,
        "chunk_B": 138
    },
    "KK": {
        "share": 0.15,
        "question": 54,
        "answer": 253,
        "chunk_A": 529,
        "chunk_B": 147
    }
}

def annual_cost(config, k):
    daily_input_tokens = 0
    daily_output_tokens = 0

    for lang, values in data.items():
        requests = requests_per_day * values["share"]
        chunk_tokens = values[f"chunk_{config}"]

        input_per_request = (
            system_prompt_tokens
            + values["question"]
            + k * chunk_tokens
        )

        output_per_request = values["answer"]

        daily_input_tokens += requests * input_per_request
        daily_output_tokens += requests * output_per_request

    annual_input_tokens = daily_input_tokens * days_per_year
    annual_output_tokens = daily_output_tokens * days_per_year

    cost = (
        annual_input_tokens * input_price
        + annual_output_tokens * output_price
    ) / 1_000_000

    return cost


results = {
    "A, k=3": annual_cost("A", 3),
    "A, k=5": annual_cost("A", 5),
    "B, k=3": annual_cost("B", 3),
    "B, k=5": annual_cost("B", 5)
}

for candidate, cost in results.items():
    print(f"{candidate}: ${cost:,.2f}")
```

```python
import pandas as pd
from math import sqrt

data = docs.copy()

data.head()
```

```python
data["accuracy"] = data["n_correct"] / data["n_docs"]

data[["pipeline", "lang", "condition", "accuracy"]]
```

```python
def wilson_interval(correct, total):
    z = 1.96
    p = correct / total

    denominator = 1 + z**2 / total

    center = (
        p + z**2 / (2 * total)
    ) / denominator

    margin = z * sqrt(
        p * (1 - p) / total
        + z**2 / (4 * total**2)
    ) / denominator

    return center - margin, center + margin
```

```python
data[["lower", "upper"]] = data.apply(
    lambda row: pd.Series(
        wilson_interval(
            row["n_correct"],
            row["n_docs"]
        )
    ),
    axis=1
)

data[[
    "pipeline", "lang", "condition",
    "accuracy", "lower", "upper"
]].round(4)
```

```python
overall = data.groupby("pipeline").agg(
    total_correct=("n_correct", "sum"),
    total_docs=("n_docs", "sum")
)

overall["overall_accuracy"] = (
    overall["total_correct"] / overall["total_docs"]
)

overall
```

```python

weakest_rows = data.loc[
    data.groupby("pipeline")["accuracy"].idxmin()
].copy()

weakest_rows[[
    "pipeline", "lang", "condition", "accuracy"
]]
```

```python
for pipeline in ["H", "L"]:
    overall_accuracy = overall.loc[
        pipeline, "overall_accuracy"
    ]

    row = weakest_rows[
        weakest_rows["pipeline"] == pipeline
    ].iloc[0]

    print(f"\nPipeline {pipeline}")
    print(f"Overall accuracy: {overall_accuracy:.2%}")
    print(f"Weakest cell: {row['lang']}, {row['condition']}")
    print(f"Accuracy there: {row['accuracy']:.2%}")
    print(
        f"95% Wilson interval: "
        f"{row['lower']:.2%} - {row['upper']:.2%}"
    )
```

```python
scores = [4.2, 2.9, 1.8, 1.2, 0.1]

exp_scores = [math.exp(x) for x in scores]
total = sum(exp_scores)
probs = [x / total for x in exp_scores]

print(probs)
```

```python
top_p = 0.95

sorted_probs = sorted(probs, reverse=True)
selected = []
cumulative = 0

for p in sorted_probs:
    selected.append(p)
    cumulative += p
    if cumulative >= top_p:
        break

new_probability = probs[0] / sum(selected)

print("Top token probability:", round(new_probability * 100, 1), "%")
```
